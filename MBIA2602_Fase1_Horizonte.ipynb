```python
%pip -q install nltk

import unicodedata
from nltk.tokenize import RegexpTokenizer

tokenizador = RegexpTokenizer(r"\w+")

def normalizar(texto):
    texto = texto.lower()
    texto = unicodedata.normalize("NFD", texto)
    texto = "".join(
        letra for letra in texto
        if unicodedata.category(letra) != "Mn"
    )
    palabras = tokenizador.tokenize(texto)
    return " ".join(palabras)

print(normalizar("¿QUÉ documentos necesito?"))
```

```python
intenciones = {
    "saludo": {
        "reglas": ["hola", "buenos dias", "buenas tardes"],
        "respuesta": "Hola, bienvenido a Universidad Horizonte. Puedo orientarte sobre inscripciones, documentos, becas, pagos, calendario y servicios."
    },
    "inscripciones": {
        "reglas": ["inscripcion", "inscribirme", "como me inscribo"],
        "respuesta": "Para inscribirte, completa el formulario de admision y entrega tu documentacion en Servicios Escolares."
    },
    "documentacion": {
        "reglas": ["documentos", "documentacion", "que papeles necesito"],
        "respuesta": "Necesitas acta de nacimiento, CURP, certificado de bachillerato y comprobante de domicilio."
    },
    "becas": {
        "reglas": ["beca", "becas", "apoyo economico"],
        "respuesta": "Universidad Horizonte ofrece becas academicas. Puedes solicitar informacion en el departamento de Becas."
    },
    "pagos": {
        "reglas": ["pago", "pagos", "colegiatura", "cuanto cuesta"],
        "respuesta": "Puedes realizar tus pagos mediante transferencia bancaria o directamente en caja."
    },
    "calendario": {
        "reglas": ["calendario", "inicio de clases", "cuando empiezan las clases"],
        "respuesta": "El proximo ciclo escolar comienza el 11 de enero de 2027."
    },
    "servicios": {
        "reglas": ["servicios", "biblioteca", "instalaciones"],
        "respuesta": "La universidad cuenta con biblioteca, laboratorios de computo, cafeteria y areas deportivas."
    },
    "despedida": {
        "reglas": ["adios", "hasta luego", "gracias"],
        "respuesta": "Gracias por consultar. Te deseamos mucho exito en tu ingreso."
    }
}

print("Intenciones cargadas:", len(intenciones))
```

```python
def responder(mensaje):
    texto = normalizar(mensaje)

    if texto == "salir":
        return "salir", "Sesion finalizada."

    coincidencias = []

    for nombre, datos in intenciones.items():
        for regla in datos["reglas"]:
            patron = " " + normalizar(regla) + " "

            if patron in " " + texto + " ":
                coincidencias.append(nombre)
                break

    if len(coincidencias) == 1:
        nombre = coincidencias[0]
        return nombre, intenciones[nombre]["respuesta"]

    if len(coincidencias) > 1:
        return "ambigua", "Por favor, consulta un solo tema por mensaje."

    return "desconocida", (
        "No entendi tu consulta. Puedes preguntar por "
        "inscripciones, documentos, becas, pagos, calendario o servicios."
    )

print(responder("¿Ofrecen becas?"))
```

```python
print("Universidad Horizonte - Asistente de Nuevo Ingreso")
print("Escribe 'salir' para terminar.")

while True:
    mensaje = input("Tu: ")
    intencion, respuesta = responder(mensaje)

    print("Intencion:", intencion)
    print("Bot:", respuesta)

    if intencion == "salir":
        break
```

```python
pruebas = [
    ("Hola", "saludo"),
    ("¿Cómo me inscribo?", "inscripciones"),
    ("¿Qué documentos necesito?", "documentacion"),
    ("¿Ofrecen becas?", "becas"),
    ("¿Cómo realizo mi pago?", "pagos"),
    ("¿Cuándo empiezan las clases?", "calendario"),
    ("¿Tienen biblioteca?", "servicios"),
    ("Adiós", "despedida"),
    ("¿Venden computadoras?", "desconocida"),
    ("becas y pagos", "ambigua"),
    ("", "desconocida"),
    ("salir", "salir")
]

aciertos = 0

for mensaje, esperada in pruebas:
    detectada, respuesta = responder(mensaje)
    correcto = detectada == esperada
    aciertos += int(correcto)

    print("Entrada:", repr(mensaje))
    print("Esperada:", esperada)
    print("Detectada:", detectada)
    print("Bot:", respuesta)
    print("Cumple:", correcto)
    print("-" * 40)

print("Coincidencias:", aciertos, "/", len(pruebas))
```
